````python
import json
import urllib.request


def agent(instruction, target_cell_code=""):
    """Local Python Code Agent powered directly by Ollama."""
    url = "http://localhost:11434/api/generate"
    system_prompt = (
        "You are an expert data analysis agent specialized in Python, Pandas, and Jupyter notebooks. "
        "Review the target code and output ONLY executable Python code blocks based on the instructions. "
        "Do not include any conversational filler."
    )

    full_prompt = f"{system_prompt}\n\nExisting Code:\n{target_cell_code}\n\nTask: {instruction}"

    payload = {
        "model": "qwen2.5-coder:1.5b",
        "prompt": full_prompt,
        "stream": False,
    }
    req = urllib.request.Request(
        url,
        data=json.dumps(payload).encode("utf-8"),
        headers={"Content-Type": "application/json"},
    )
    try:
        with urllib.request.urlopen(req) as response:
            result = json.loads(response.read().decode("utf-8"))
            print(result["response"].replace("```python", "").replace("```", ""))
    except Exception as e:
        print(f"Error communicating with local server: {e}")
````

````python
agent("What is your model name and version?")
````
